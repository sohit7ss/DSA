# Count Number of Bits to Be Flipped

Count the number of bits that need to be flipped to convert **A → B**.

### Example 1

**Start = 10**  
**Goal = 7**

```text
10 → 1010
 7 → 0111

 1010 → 0111

 Answer = 3

````python
def count_bit_of_flip(start, goal):
    ans = start ^ goal 
    count = 0
    for i in range (0,32):
        if ans &(1<<i) != 0:
            count += 1
    return count

# TC = O(32)
# SC = O(1)
````

````python
print(count_bit_of_flip(10,7))
````
